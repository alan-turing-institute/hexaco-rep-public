```python
def plot_for_paper(values, title, n_display = 15):
    """
    Plot the eigenvalues for a PCA analysis, for use in the paper.
    """
    plt.figure(figsize=(4, 2.4)) 
    plt.rc('font', family='Charter')
    plt.plot(values[:n_display], '.-', linewidth=1, markersize=8, alpha=0.6, label=title) 
    
    text_size = 8
    plt.xlabel('Component Number', fontsize=text_size) 
    plt.ylabel('Eigenvalue', fontsize=text_size) 
    plt.gca().yaxis.set_tick_params(labelsize=text_size-1)
    plt.gca().xaxis.set_tick_params(labelsize=text_size-1)
    l = plt.legend()
    plt.setp(l.texts, fontsize=text_size)
    plt.grid(True)  
    plt.show()
    
plot_for_paper(eigenvalues_b, "PCA(unrotated) for PopCensus", n_display=15)
```
