# Cronbach's Alpha across all Solutions

The alpha value is calculated for all factors within a solution, 
and the mean average is then used to determine the optimal solution.

```python
from os.path import exists
import pandas as pd
import numpy as np
from support.solution_support import *

# 'p' for professional population, 'c' for census population
df_census = plot_cronbachs_for_population('c')
plot_cronbachs_for_population('p')
```

```python
def plot_for_paper(df, xlabel, ylabel):
    import matplotlib.pyplot as plt
    import seaborn as sns
   
    plt.figure(figsize=(5, 2.8))
    plt.rc('font', family='Times New Roman')

    textsize = 8
    sns.heatmap(
        df,
        annot=True,
        fmt=".2f",
        cmap=plt.cm.RdBu,
        alpha=0.6,
        linewidths=0.7,
        annot_kws={"fontsize": textsize, "color": "black"},
        vmin=-1.5,
        vmax=1.5,
        cbar=False
    )

    plt.xlabel(xlabel, fontsize=textsize+1)
    plt.xticks(rotation=0, fontsize=textsize)
    plt.ylabel(ylabel, fontsize=textsize+1)
    plt.yticks(rotation=0, fontsize=textsize)

    plt.tight_layout()
    plt.show()

plot_for_paper(df_census, "Solutions (mean alpha) - PopCensus", "Factors")
```
