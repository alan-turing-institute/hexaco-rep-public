```python
from results_template import generate_results
from support.pca_support import popc_10_factor_labels as factor_labels

internal_sims, hexaco_sims = generate_results(10, 'PopCensus', 
                                              'data/popc_ipsatised_results.csv', 
                                              'intermediate/loadings_popc_10.csv',
                                              factor_labels=factor_labels
                                              )
```

```python
def plot_internal_sims_for_paper(data, n_factors):
    import matplotlib.pyplot as plt
    import numpy as np
    
    all_means = data['means']
    all_errors = data['stds']
    labels = factor_labels

    all_labels = [f"{x}. {labels[x-1]}" for x in range(1, 11)] + list(data.index)[10:]
    
    import seaborn as sns
    sns.set_theme(style='white', font='Times New Roman')
    colors = sns.color_palette("tab20", n_colors=20)
    palette = [colors[0], colors[2], colors[4]]
    bar_colors = ([palette[0]]*n_factors) + ([palette[2]]*1) + ([palette[1]]*6)


    
    import matplotlib.patches as mpatches
    p1 = mpatches.Patch(color=palette[0], alpha=0.7, label='Recovered Factors')
    p2 = mpatches.Patch(color=palette[1], alpha=0.7, label='HEXACO Dimensions')
    p3 = mpatches.Patch(color=palette[2], alpha=0.7, label='Random Baseline')

    textsize = 9
    
    pos = np.arange(len(all_means)) 
    fig = plt.figure(figsize=(4, 3.5))
    plt.barh(pos, all_means, xerr=all_errors, capsize=3, alpha=0.7, color=bar_colors)
    plt.yticks(pos, all_labels, rotation=0, ha='right', fontsize=textsize)
    plt.xticks(fontsize=textsize)
    plt.gca().spines['top'].set_visible(False)
    plt.gca().spines['right'].set_visible(False)
    plt.gca().spines['left'].set_linewidth(1)
    plt.gca().spines['bottom'].set_linewidth(1)
    plt.axvline(0, color='black', linestyle='-', linewidth=1)
    plt.xlim(0, 0.6)
    fig.gca().invert_yaxis()
    plt.xlabel("Semantic Similarity Score", fontsize=textsize)
    plt.ylabel("Factors/Dimensions", fontsize=textsize)
    plt.legend(handles=[p1, p3, p2], 
               bbox_to_anchor=(0.25, -0.2), 
               loc='upper center', 
               fontsize=textsize, 
               ncol=3, 
               title_fontsize=textsize)
    plt.show()

# fix the order of the columns, and then plot it!
order = ['Factor1', 'Factor2', 'Factor3', 'Factor4', 'Factor5', 
         'Factor6', 'Factor7', 'Factor8', 'Factor9', 'Factor10', 
         'Random', 
         'Honesty-Humility', 'Emotionality', 'Extraversion',
         'Agreeableness', 'Conscientiousness', 'Openness']
internal_sims = internal_sims.reindex(order)
plot_internal_sims_for_paper(internal_sims, 10)
```

```python

def plot_hexaco_sims_for_paper(df):
    import matplotlib.patches as mpatches
    import matplotlib.pyplot as plt
    import seaborn as sns
    import numpy as np
    
    sns.set_theme(style='white', font='Times New Roman')
    
    colors = sns.color_palette("tab20", n_colors=20)
    colors = [colors[0], colors[1], colors[2], colors[3],
              colors[4], colors[5]]
    colors.reverse()
    
    labels = factor_labels
    labels = [f"{x}. {labels[x-1]}" for x in range(1, 11)]
    labels.reverse()
    
    textsize = 9

    fig, ax = plt.subplots(figsize=(4.5, 4))

    df.iloc[::-1].plot(kind='barh', ax=ax, width=0.8, color=colors, linewidth=0, alpha=0.8)   
    plt.gca().spines['top'].set_visible(False)
    plt.gca().spines['right'].set_visible(False)
    plt.gca().spines['left'].set_linewidth(1)
    plt.gca().spines['bottom'].set_linewidth(1)
    plt.xlabel('Semantic Similarity Score', fontsize=textsize)
    plt.ylabel('Factor', fontsize=textsize)
    plt.xticks(rotation=0, fontsize=textsize)
    plt.xlim(0, 0.6)
    plt.yticks(np.arange(len(labels)), labels, fontsize=textsize)
    handles, labels = ax.get_legend_handles_labels()
    ax.legend(
        handles[::-1],
        labels[::-1],
        bbox_to_anchor=(0.25, -0.2), 
        loc='upper center',
        fontsize=textsize, 
        ncol=3
    )
    plt.tight_layout()
    plt.show()

hexaco_order = ['Openness', 'Conscientiousness', 'Agreeableness', 'Extraversion', 'Emotionality', 'Honesty-Humility']
hexaco_sims = hexaco_sims[hexaco_order]
plot_hexaco_sims_for_paper(hexaco_sims)
```
