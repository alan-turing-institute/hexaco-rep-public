### Calculate Semantic Similairty of factors between PopCensus' 10 Factor solution, and PopProfessional's 8 Factor Solution

To see how similar the different populations' recovered personality structures are.

```python
from support.semantic_similarity import compare_solutions
from support.pca_support import popc_10_factor_labels, popp_8_factor_labels
from support.semantic_similarity import load_embedding_model

df = compare_solutions("intermediate/loadings_popc_10.csv", popc_10_factor_labels, 
                       "intermediate/loadings_popp_8.csv", popp_8_factor_labels, 
                       load_embedding_model())
df
```

```python
def plot_for_paper(df):
    import pandas as pd
    import seaborn as sns
    import matplotlib.pyplot as plt

    sns.set_theme(style='white', font='Times New Roman')

    data = df
    data = data.apply(pd.to_numeric)
    data = data

    #fig = plt.figure(figsize=(10, 5))
    ax = sns.heatmap(data, annot=True, cmap='Blues', fmt=".2f", linewidths=0.5, vmin=0, vmax=1, cbar=False, annot_kws={"fontsize":10})

    plt.gca().invert_yaxis()
    ax.set_yticklabels([f"{i+1}. {x}" for i, x in enumerate(popp_8_factor_labels)], rotation=0, size=10)
    plt.xlabel('\n10-Factor solution from PopCensus', size=12)
    plt.ylabel('8-Factor solution from PopProfessional\n', size=12)
    ax.set_xticklabels([f"{i+1}. {x}" for i, x in enumerate(popc_10_factor_labels)], rotation=90, size=10)

    plt.show()

plot_for_paper(df)
```
